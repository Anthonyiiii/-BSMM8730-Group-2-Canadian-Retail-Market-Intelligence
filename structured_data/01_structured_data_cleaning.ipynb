```python
import pandas as pd

file_path = "/Users/yuanquan0901/Desktop/BSMM 8730/group project/20100056.csv"

df = pd.read_csv(file_path)

print("Shape:", df.shape)

df.head()
```

```python
print(df.columns.tolist())

print("\nData types:")

print(df.dtypes)

print("\nMissing values:")

print(df.isnull().sum())

print("\nUnique values:")

print(df.nunique())
```

```python
print("Date range:")

print(df["REF_DATE"].min(), "to", df["REF_DATE"].max())

print("\nGeographies:")

print(df["GEO"].unique())

print("\nSales types:")

print(df["Sales"].value_counts())

print("\nAdjustments:")

print(df["Adjustments"].value_counts())

print("\nUnits:")

print(df["UOM"].value_counts())

print("\nScalar factors:")

print(df["SCALAR_FACTOR"].value_counts())

print("\nStatus:")

print(df["STATUS"].value_counts(dropna=False))
```

```python
missing_value_rows = df[df["VALUE"].isna()]

print("Rows with missing VALUE:", len(missing_value_rows))

print("\nMissing VALUE by status:")

print(

    missing_value_rows["STATUS"]

    .value_counts(dropna=False)

)

print("\nMissing VALUE by sales type:")

print(

    missing_value_rows["Sales"]

    .value_counts()

)

print("\nMissing VALUE by adjustment:")

print(

    missing_value_rows["Adjustments"]

    .value_counts()

)

print("\nMissing VALUE by geography:")

print(

    missing_value_rows["GEO"]

    .value_counts()

)
```

```python
df_clean = df.copy()

print("Original shape:", df.shape)

print("Working copy shape:", df_clean.shape)
```

```python
df_clean = df_clean.rename(columns={

    "REF_DATE": "date",

    "GEO": "geography",

    "DGUID": "dguid",

    "North American Industry Classification System (NAICS)": "industry",

    "Sales": "sales_type",

    "Adjustments": "adjustment",

    "UOM": "unit",

    "SCALAR_FACTOR": "scalar_factor",

    "VALUE": "sales_value",

    "STATUS": "status"

})

print(df_clean.columns.tolist())
```

```python
df_clean["date"] = pd.to_datetime(

    df_clean["date"],

    format="%Y-%m"

)

df_clean["year"] = df_clean["date"].dt.year

df_clean["month"] = df_clean["date"].dt.month

print(df_clean["date"].dtype)

print(df_clean["date"].min())

print(df_clean["date"].max())
```

```python
columns_to_drop = [
    "UOM_ID",
    "SCALAR_ID",
    "VECTOR",
    "COORDINATE",
    "SYMBOL",
    "TERMINATED",
    "DECIMALS"
]

df_clean = df_clean.drop(columns=columns_to_drop)

print(df_clean.shape)
print(df_clean.columns.tolist())
```

```python
print("Before removing missing sales values:", len(df_clean))

df_clean = df_clean.dropna(

    subset=["sales_value"]

).copy()

print("After removing missing sales values:", len(df_clean))

print(

    "Remaining missing sales values:",

    df_clean["sales_value"].isna().sum()

)
```

```python
print("Final shape so far:", df_clean.shape)

print("\nMissing values:")

print(df_clean.isnull().sum())

print("\nData types:")

print(df_clean.dtypes)

df_clean.head()
```

```python
print("Exact duplicate rows:", df_clean.duplicated().sum())

key_columns = [

    "date",

    "geography",

    "industry",

    "sales_type",

    "adjustment"

]

print(

    "Duplicate business keys:",

    df_clean.duplicated(subset=key_columns).sum()

)

print("\nSales types:")

print(df_clean["sales_type"].value_counts())

print("\nAdjustments:")

print(df_clean["adjustment"].value_counts())

print("\nUnits:")

print(df_clean["unit"].value_counts())

print("\nScalar factors:")

print(df_clean["scalar_factor"].value_counts())
```

```python
ecommerce = df_clean[
    df_clean["sales_type"] == "Retail e-commerce sales"
].copy()

print("E-commerce rows:", len(ecommerce))

print("\nDate range:")
print(ecommerce["date"].min(), "to", ecommerce["date"].max())

print("\nGeographies:")
print(ecommerce["geography"].value_counts())

print("\nIndustries:")
print(ecommerce["industry"].value_counts())

print("\nAdjustments:")
print(ecommerce["adjustment"].value_counts())
```

```python
df_clean["sales_millions"] = (

    df_clean["sales_value"] / 1000

).round(2)

df_clean["sales_billions"] = (

    df_clean["sales_value"] / 1000000

).round(4)

df_clean[

    [

        "date",

        "geography",

        "industry",

        "sales_type",

        "sales_value",

        "sales_millions",

        "sales_billions"

    ]

].head()
```

```python
print("Number of industries:", df_clean["industry"].nunique())

print("\nIndustries:")

for industry in sorted(df_clean["industry"].unique()):

    print(industry)
```

```python
print("Number of geographies:", df_clean["geography"].nunique())

print("\nRows by geography:")

print(

    df_clean["geography"]

    .value_counts()

)
```

```python
provinces_territories = [

    "Newfoundland and Labrador",

    "Prince Edward Island",

    "Nova Scotia",

    "New Brunswick",

    "Quebec",

    "Ontario",

    "Manitoba",

    "Saskatchewan",

    "Alberta",

    "British Columbia",

    "Yukon",

    "Northwest Territories",

    "Nunavut"

]

df_clean["geography_level"] = "City/CMA"

df_clean.loc[

    df_clean["geography"].isin(provinces_territories),

    "geography_level"

] = "Province/Territory"

df_clean.loc[

    df_clean["geography"] == "Canada",

    "geography_level"

] = "Canada"

print(df_clean["geography_level"].value_counts())

print(

    df_clean[

        ["geography", "geography_level"]

    ].drop_duplicates().sort_values(

        ["geography_level", "geography"]

    )

)
```

```python
df_clean["naics_code"] = (

    df_clean["industry"]

    .str.extract(r"\[([^\]]+)\]")[0]

)

print(

    df_clean[

        ["industry", "naics_code"]

    ].drop_duplicates().sort_values("naics_code")

)
```

```python
df_clean["status"] = df_clean["status"].fillna("No flag")

print(df_clean["status"].value_counts(dropna=False))

print("\nMissing status values:")

print(df_clean["status"].isna().sum())
```

```python
print("Final shape:", df_clean.shape)

print("\nMissing values:")

print(df_clean.isnull().sum())

print("\nExact duplicate rows:")

print(df_clean.duplicated().sum())

print("\nDate range:")

print(df_clean["date"].min(), "to", df_clean["date"].max())

print("\nGeographies:", df_clean["geography"].nunique())

print("Industries:", df_clean["industry"].nunique())

print("\nSales types:")

print(df_clean["sales_type"].value_counts())

print("\nGeography levels:")

print(df_clean["geography_level"].value_counts())

print("\nFinal columns:")

print(df_clean.columns.tolist())
```

```python
output_path = "/Users/yuanquan0901/Desktop/BSMM 8730/group project/retail_sales_clean.csv"

df_clean.to_csv(

    output_path,

    index=False,

    encoding="utf-8-sig"

)

print("Saved to:")

print(output_path)
```
