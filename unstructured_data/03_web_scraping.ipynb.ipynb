```python
import requests

import pandas as pd

from bs4 import BeautifulSoup

import re

def scrape_article(url):

    # Set request headers

    headers = {

        "User-Agent": "Mozilla/5.0"

    }

    # Request the webpage

    response = requests.get(url, headers=headers, timeout=20)

    response.raise_for_status()

    # Parse HTML

    soup = BeautifulSoup(response.text, "html.parser")

    main = soup.find("main")

    if main is None:

        return None

    # Extract page title

    h1 = main.find("h1")

    title = h1.get_text(" ", strip=True) if h1 else "Unknown"

    # Extract publication date

    page_text = main.get_text(" ", strip=True)

    date_match = re.search(

        r"(January|February|March|April|May|June|July|August|September|October|November|December)"

        r"\s+\d{1,2},\s+\d{4}",

        page_text

    )

    publication_date = date_match.group(0) if date_match else None

    # Find the beginning of the main article content

    start_heading = None

    for h in main.find_all(["h2", "h3"]):

        heading_text = h.get_text(" ", strip=True)

        if heading_text in [

            "Executive summary",

            "Overview",

            "Introduction"

        ]:

            start_heading = h

            break

    # Select elements to extract

    if start_heading:

        elements = start_heading.find_all_next()

    else:

        elements = main.find_all(["h2", "h3", "p"])

    # Extract article content

    content = []

    for element in elements:

        text = element.get_text(" ", strip=True)

        if (

            element.name == "h2"

            and text in [

                "How to contact the Competition Bureau",

                "Footnotes"

            ]

        ):

            break

        if element.name in ["h2", "h3", "p"] and text:

            content.append(text)

    # Combine and clean text

    article_text = " ".join(content)

    article_text = re.sub(r"\s+", " ", article_text).strip()

    # Return extracted data

    return {

        "title": title,

        "publication_date": publication_date,

        "source": "Competition Bureau Canada",

        "source_url": url,

        "article_text": article_text,

        "word_count": len(article_text.split())

    }

# Test the scraping function

url = "https://competition-bureau.canada.ca/en/how-we-foster-competition/education-and-outreach/canada-needs-more-grocery-competition"

result = scrape_article(url)

print("Title:", result["title"])

print("Date:", result["publication_date"])

print("Word Count:", result["word_count"])

print()

print(result["article_text"][:1000])
```

```python
def scrape_article(url):

    headers = {

        "User-Agent": "Mozilla/5.0"

    }

    response = requests.get(url, headers=headers, timeout=20)

    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    main = soup.find("main")

    if main is None:

        return None

    # Extract title

    h1 = main.find("h1")

    title = h1.get_text(" ", strip=True) if h1 else "Unknown"

    # Extract publication date

    page_text = main.get_text(" ", strip=True)

    date_match = re.search(

        r"(January|February|March|April|May|June|July|August|September|October|November|December)"

        r"\s+\d{1,2},\s+\d{4}",

        page_text

    )

    publication_date = date_match.group(0) if date_match else None

    # Remove unnecessary HTML elements

    for tag in main.find_all(["script", "style", "nav", "form", "button"]):

        tag.decompose()

    # Identify possible content starting points

    start_headings = [

        "Executive summary",

        "Notice of Study",

        "Overview",

        "Introduction"

    ]

    start_element = None

    for element in main.find_all(["h2", "h3"]):

        text = element.get_text(" ", strip=True)

        if text in start_headings:

            start_element = element

            break

    # Extract content

    if start_element:

        elements = start_element.find_all_next(["h2", "h3", "p", "li"])

    else:

        elements = main.find_all(["h2", "h3", "p", "li"])

    content = []

    for element in elements:

        text = element.get_text(" ", strip=True)

        if not text:

            continue

        if element.name == "h2" and text in [

            "How to contact the Competition Bureau",

            "Footnotes"

        ]:

            break

        content.append(text)

    # Remove duplicate consecutive elements

    cleaned_content = []

    for text in content:

        if not cleaned_content or text != cleaned_content[-1]:

            cleaned_content.append(text)

    # Combine and clean text

    article_text = " ".join(cleaned_content)

    article_text = re.sub(r"\s+", " ", article_text).strip()

    return {

        "title": title,

        "publication_date": publication_date,

        "source": "Competition Bureau Canada",

        "source_url": url,

        "article_text": article_text,

        "word_count": len(article_text.split())

    }
```
