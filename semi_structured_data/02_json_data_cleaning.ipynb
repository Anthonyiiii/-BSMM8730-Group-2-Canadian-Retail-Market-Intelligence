```python
import json

import pandas as pd

import os

file_path = "/Users/yuanquan0901/Desktop/BSMM 8730/group project/ind-econ_(semi-structured data).json"

with open(file_path, "r", encoding="utf-8") as f:

    data = json.load(f)

print("JSON loaded successfully")

print(type(data))

print(data.keys())
```

```python
print(type(data["results"]))

print("\nKeys inside results:")

print(data["results"].keys())
```

```python
indicators = data["results"]["indicators"]

print("Type:", type(indicators))

print("Number of indicators:", len(indicators))

print("\nFirst indicator:")

print(indicators[0])
```

```python
df_json = pd.json_normalize(indicators)

print("Shape:", df_json.shape)

print("\nColumns:")

for column in df_json.columns:

    print(column)
```

```python
indicator_list = df_json[

    [

        "indicator_number",

        "geo_code",

        "title.en",

        "value.en",

        "refper.en",

        "source",

        "release_date",

        "growth_rate.growth.en",

        "growth_rate.details.en"

    ]

].copy()

pd.set_option("display.max_rows", 250)

pd.set_option("display.max_colwidth", 100)

indicator_list
```

```python
keywords = [

    "retail",

    "earnings",

    "consumer",

    "price",

    "employment",

    "unemployment",

    "population",

    "GDP"

]

pattern = "|".join(keywords)

relevant = indicator_list[

    indicator_list["title.en"].str.contains(

        pattern,

        case=False,

        na=False

    )

]

print("Potentially relevant indicators:", len(relevant))

relevant
```

```python
selected_indicators = [

    "Average weekly earnings",

    "Consumer Price Index",

    "Employment level",

    "Unemployment rate",

    "Real GDP by expenditure"

]

df_econ = df_json[

    df_json["title.en"].isin(selected_indicators)

].copy()

print("Rows selected:", len(df_econ))

print("\nIndicators:")

print(df_econ["title.en"].value_counts())

df_econ.head()
```

```python
df_econ = df_econ[

    [

        "indicator_number",

        "geo_code",

        "title.en",

        "value.en",

        "refper.en",

        "source",

        "release_date",

        "growth_rate.growth.en",

        "growth_rate.details.en"

    ]

].copy()

df_econ.columns = [

    "indicator_number",

    "geo_code",

    "indicator",

    "value",

    "reference_period",

    "source",

    "release_date",

    "growth_rate",

    "growth_details"

]

print(df_econ.shape)

df_econ.head()
```

```python
geo_data = data["results"]["geo"]

print(type(geo_data))

print("Number of geographies:", len(geo_data))

geo_data[:5]
```

```python
for item in geo_data[:5]:

    print(item)
```

```python
geo_map = {

    int(item["geo_code"]): item["label"]["en"]

    for item in geo_data

}

print(geo_map)
```

```python
df_econ["geography"] = df_econ["geo_code"].map(geo_map)

print(df_econ[["geo_code", "geography"]].drop_duplicates().sort_values("geo_code"))
```

```python
print("Missing geography:", df_econ["geography"].isna().sum())

print("\nRows by geography:")

print(df_econ["geography"].value_counts())
```

```python
df_econ["value_original"] = df_econ["value"]

df_econ["numeric_value"] = (

    df_econ["value"]

    .astype(str)

    .str.replace("$", "", regex=False)

    .str.replace(",", "", regex=False)

    .str.replace("%", "", regex=False)

    .str.strip()

)

df_econ["numeric_value"] = pd.to_numeric(

    df_econ["numeric_value"],

    errors="coerce"

)

df_econ[

    ["indicator", "geography", "value_original", "numeric_value"]

].head(20)
```

```python
unit_map = {

    "Average weekly earnings": "Dollars",

    "Consumer Price Index": "Index",

    "Employment level": "Persons",

    "Unemployment rate": "Percent",

    "Real GDP by expenditure": "Index"

}

df_econ["unit"] = df_econ["indicator"].map(unit_map)

df_econ[

    ["indicator", "value_original", "numeric_value", "unit"]

].drop_duplicates()
```

```python
df_econ["growth_numeric"] = (

    df_econ["growth_rate"]

    .astype(str)

    .str.extract(r"([-+]?\d*\.?\d+)")[0]

)

df_econ["growth_numeric"] = pd.to_numeric(

    df_econ["growth_numeric"],

    errors="coerce"

)

df_econ[

    [

        "indicator",

        "geography",

        "growth_rate",

        "growth_numeric",

        "growth_details"

    ]

].head(20)
```

```python
df_econ["release_date"] = pd.to_datetime(

    df_econ["release_date"],

    errors="coerce"

)

print(df_econ["release_date"].dtype)

print(df_econ["release_date"].min())

print(df_econ["release_date"].max())
```

```python
print("Shape:", df_econ.shape)

print("\nMissing values:")

print(df_econ.isna().sum())

print("\nDuplicate rows:")

print(df_econ.duplicated().sum())

print("\nRows by indicator:")

print(df_econ["indicator"].value_counts())
```

```python
missing_numeric = df_econ[

    df_econ["numeric_value"].isna()

][

    [

        "indicator",

        "geography",

        "value_original",

        "numeric_value",

        "growth_rate",

        "growth_numeric",

        "reference_period"

    ]

]

missing_numeric
```

```python
final_columns = [

    "indicator_number",

    "indicator",

    "geo_code",

    "geography",

    "value_original",

    "numeric_value",

    "reference_period",

    "release_date",

    "growth_rate",

    "growth_numeric",

    "growth_details",

    "source"

]

df_econ_clean = df_econ[final_columns].copy()

df_econ_clean.head()
```

```python
print("Final shape:", df_econ_clean.shape)

print("\nMissing values:")

print(df_econ_clean.isna().sum())

print("\nDuplicate rows:")

print(df_econ_clean.duplicated().sum())

print("\nIndicators:")

print(df_econ_clean["indicator"].value_counts())

print("\nGeographies:")

print(df_econ_clean["geography"].value_counts())

print("\nData types:")

print(df_econ_clean.dtypes)
```

```python
df_econ_clean.to_csv(

    "economic_indicators_clean.csv",

    index=False,

    encoding="utf-8-sig"

)

print("File exported successfully")

print(df_econ_clean.shape)
```
